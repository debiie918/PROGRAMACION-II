Escribe una función que tome una lista de números como argumento
y devuelva la suma de los números positivos y la suma de los
números negativos en una tupla.

```python
positivos_negativos = (10,-5,50,-10,30,-25,60,80)
negativos=[]
positivos=[]
for i in positivos_negativos:
    if i < 0:
        negativos.append(i)
    else:
        positivos.append(i)
print(f"La cantida de negativos es igual a {len(negativos)}")
print(f"La cantida de positivos es igual a {len(positivos)}")
        
```
